```python
# ===================================================================
# ANÁLISE POR PAÍS - AMÉRICAS
# ===================================================================

print("🌍 ANÁLISE POR PAÍS DAS AMÉRICAS")
print("=" * 70)

# Verificar se há dados
if len(df) == 0:
    print("❌ DataFrame vazio! Carregue os dados primeiro.")
    raise ValueError("DataFrame vazio")

# ===================================================================
# TOP PAÍSES DAS AMÉRICAS
# ===================================================================

# Top países por streams
top_paises = (df.groupby('country')
              .agg({
                  'streams_millions': 'sum',
                  'track_name': 'count'
              })
              .rename(columns={'track_name': 'total_registros'})
              .sort_values('streams_millions', ascending=False)
              .head(15)
              .reset_index())

# Adicionar sub-região se disponível
if 'sub_region' in df.columns:
    regiao_map = df.groupby('country')['sub_region'].first().to_dict()
    top_paises['sub_region'] = top_paises['country'].map(regiao_map)

print("\n🏆 TOP 10 PAÍSES DAS AMÉRICAS POR STREAMS:")
print("-" * 70)
for i, row in top_paises.head(10).iterrows():
    pct = (row['streams_millions'] / top_paises['streams_millions'].sum()) * 100
    if 'sub_region' in top_paises.columns:
        print(f"{i+1:2}. {row['country']:22} : {row['streams_millions']:8.1f}M streams ({pct:.1f}%) - {row['sub_region']}")
    else:
        print(f"{i+1:2}. {row['country']:22} : {row['streams_millions']:8.1f}M streams ({pct:.1f}%)")

# ===================================================================
# GRÁFICO DE BARRAS - TOP PAÍSES DAS AMÉRICAS
# ===================================================================

# Cores baseadas em sub-região
if 'sub_region' in top_paises.columns:
    cores_regioes = {
        '🌎 América do Norte': '#4ECDC4',
        '🗺️ América do Sul': '#45B7D1',
        '🌏 América Central': '#FFE66D',
        '🏝️ Caribe': '#FF6B6B',
        '📍 Outros': '#95A5A6'
    }

    fig_top_paises = px.bar(
        top_paises,
        x='country',
        y='streams_millions',
        color='sub_region',
        text='streams_millions',
        title='🌎 Top 15 Países das Américas por Total de Streams',
        labels={'streams_millions': 'Streams (Milhões)', 'country': 'País', 'sub_region': 'Sub-região'},
        color_discrete_map=cores_regioes
    )
else:
    fig_top_paises = px.bar(
        top_paises,
        x='country',
        y='streams_millions',
        color='streams_millions',
        text='streams_millions',
        title='🌎 Top 15 Países das Américas por Total de Streams',
        labels={'streams_millions': 'Streams (Milhões)', 'country': 'País'},
        color_continuous_scale='Plasma'
    )

fig_top_paises.update_traces(texttemplate='%{text:.1f}M', textposition='outside')
fig_top_paises.update_layout(
    height=550,
    title_x=0.5,
    title_font_size=18,
    xaxis_title="",
    yaxis_title="Streams (Milhões)"
)
fig_top_paises.show()

print("✅ Gráfico 1: Top países das Américas gerado!")

# ===================================================================
# TABELA INTERATIVA DE PAÍSES
# ===================================================================

print("\n📊 TABELA INTERATIVA DE PAÍSES DAS AMÉRICAS:")

# Calcular percentuais
top_paises['percentual'] = (top_paises['streams_millions'] / top_paises['streams_millions'].sum() * 100).round(1)

# Adicionar barras de progresso para visualização
def criar_barra(pct, tamanho=20):
    barras = int((pct / 100) * tamanho)
    return '█' * barras + '░' * (tamanho - barras)

top_paises['barra'] = top_paises['percentual'].apply(lambda x: criar_barra(x, 20))

# Tabela interativa
fig_tabela_paises = go.Figure(data=[go.Table(
    header=dict(
        values=['<b>País</b>', '<b>Sub-região</b>', '<b>Streams (M)</b>', '<b>Registros</b>', '<b>% Total</b>', '<b>Distribuição</b>'],
        fill_color='#2C3E50',
        font=dict(color='white', size=12),
        align='left',
        height=40
    ),
    cells=dict(
        values=[
            top_paises['country'],
            top_paises['sub_region'] if 'sub_region' in top_paises.columns else ['N/A'] * len(top_paises),
            top_paises['streams_millions'].round(1),
            top_paises['total_registros'],
            top_paises['percentual'],
            top_paises['barra']
        ],
        fill_color=['lavender', 'lavender', 'lavender', 'lavender', 'lavender', 'white'],
        align='left',
        height=35
    )
)])

fig_tabela_paises.update_layout(
    title='📊 Distribuição de Streams por País - Américas',
    height=600,
    title_x=0.5
)
fig_tabela_paises.show()

print("✅ Tabela interativa gerada!")

# ===================================================================
# ANÁLISE DE CONCENTRAÇÃO POR SUB-REGIÃO
# ===================================================================

if 'sub_region' in df.columns:
    print("\n" + "="*70)
    print("🗺️ CONCENTRAÇÃO DE STREAMS POR SUB-REGIÃO")
    print("="*70)

    # Agregar por sub-região
    streams_por_regiao = (df.groupby('sub_region')['streams_millions']
                          .sum()
                          .sort_values(ascending=False)
                          .reset_index())

    # Calcular percentuais
    total_streams = streams_por_regiao['streams_millions'].sum()
    streams_por_regiao['percentual'] = (streams_por_regiao['streams_millions'] / total_streams * 100).round(1)

    print("\n📊 Distribuição de streams por sub-região:")
    print("-" * 50)
    for _, row in streams_por_regiao.iterrows():
        barra = '█' * int(row['percentual'] / 2)
        print(f"   {row['sub_region']:20} : {row['streams_millions']:8.1f}M ({row['percentual']:.1f}%) {barra}")

    # Gráfico de pizza por região
    fig_pizza_regioes = px.pie(
        streams_por_regiao,
        values='streams_millions',
        names='sub_region',
        title='🌎 Distribuição de Streams por Sub-região nas Américas',
        hole=0.3,
        color_discrete_sequence=px.colors.qualitative.Set3
    )
    fig_pizza_regioes.update_traces(textposition='inside', textinfo='percent+label')
    fig_pizza_regioes.update_layout(height=500, title_x=0.5)
    fig_pizza_regioes.show()

    print("✅ Gráfico 2: Distribuição por sub-região gerado!")

# ===================================================================
# MÉTRICAS ADICIONAIS POR PAÍS
# ===================================================================

print("\n" + "="*70)
print("📈 MÉTRICAS ADICIONAIS POR PAÍS")
print("="*70)

# Calcular métricas por país
if 'danceability' in df.columns and 'energy' in df.columns and 'valence' in df.columns:
    metricas_paises = (df.groupby('country')
                       .agg({
                           'streams_millions': 'sum',
                           'danceability': 'mean',
                           'energy': 'mean',
                           'valence': 'mean',
                           'rank': 'mean'
                       })
                       .round(3)
                       .sort_values('streams_millions', ascending=False)
                       .head(8)
                       .reset_index())

    print("\n🎵 PERFIL MUSICAL DOS TOP 8 PAÍSES:")
    print("-" * 80)
    print(f"{'País':22} | {'Streams(M)':>10} | {'Dança':>5} | {'Energia':>5} | {'Positividade':>5} | {'Rank médio':>9}")
    print("-" * 80)

    for _, row in metricas_paises.iterrows():
        print(f"{row['country']:22} | {row['streams_millions']:8.1f}M | {row['danceability']:5.2f} | {row['energy']:5.2f} | {row['valence']:5.2f} | {row['rank']:8.1f}")

# ===================================================================
# RESUMO DA ANÁLISE POR PAÍS
# ===================================================================

print("\n" + "="*70)
print("✅ RESUMO DA ANÁLISE POR PAÍS")
print("="*70)

pais_top = top_paises.iloc[0]['country']
streams_top = top_paises.iloc[0]['streams_millions']
pct_top = (streams_top / top_paises['streams_millions'].sum()) * 100

print(f"""
🏆 PRINCIPAIS DESCOBERTAS:

   • 🇺🇸 País com mais streams: {pais_top} ({streams_top:.1f}M streams - {pct_top:.1f}% do total)
   • 🌍 Concentração: O Top 3 países representa {(top_paises.head(3)['streams_millions'].sum() / top_paises['streams_millions'].sum() * 100):.1f}% de todos os streams
   • 📊 Países analisados: {len(top_paises)} países das Américas
""")

if 'sub_region' in df.columns:
    regiao_top = streams_por_regiao.iloc[0]['sub_region']
    pct_regiao = streams_por_regiao.iloc[0]['percentual']
    print(f"   • 🗺️ Sub-região dominante: {regiao_top} ({pct_regiao:.1f}% dos streams)")

print("\n✅ Análise por país concluída!")

# ===================================================================
# COMPARAÇÃO AMÉRICA DO NORTE VS AMÉRICA DO SUL
# ===================================================================

if 'sub_region' in df.columns:
    print("\n📊 COMPARAÇÃO AMÉRICA DO NORTE VS AMÉRICA DO SUL")
    print("="*50)

    norte = df[df['sub_region'] == '🌎 América do Norte']
    sul = df[df['sub_region'] == '🗺️ América do Sul']

    comparacao = pd.DataFrame({
        'Métrica': ['Total de Streams (M)', 'Média de Streams', 'Dançabilidade', 'Energia', 'Positividade', 'Artistas Únicos'],
        'América do Norte': [
            f"{norte['streams_millions'].sum():.1f}M",
            f"{norte['streams_millions'].mean():.2f}M",
            f"{norte['danceability'].mean():.3f}" if 'danceability' in norte.columns else 'N/A',
            f"{norte['energy'].mean():.3f}" if 'energy' in norte.columns else 'N/A',
            f"{norte['valence'].mean():.3f}" if 'valence' in norte.columns else 'N/A',
            f"{norte['artist_names'].nunique():,}"
        ],
        'América do Sul': [
            f"{sul['streams_millions'].sum():.1f}M",
            f"{sul['streams_millions'].mean():.2f}M",
            f"{sul['danceability'].mean():.3f}" if 'danceability' in sul.columns else 'N/A',
            f"{sul['energy'].mean():.3f}" if 'energy' in sul.columns else 'N/A',
            f"{sul['valence'].mean():.3f}" if 'valence' in sul.columns else 'N/A',
            f"{sul['artist_names'].nunique():,}"
        ]
    })

    print(comparacao.to_string(index=False))
```

```python
# ===================================================================
# PERFIL MUSICAL POR SUB-REGIÃO DAS AMÉRICAS
# ===================================================================

print("🎸 PERFIL MUSICAL POR SUB-REGIÃO DAS AMÉRICAS")
print("=" * 70)

# Verificar se a coluna sub_region existe
if 'sub_region' not in df.columns:
    print("⚠️ Coluna 'sub_region' não encontrada!")
    print("Criando sub-regiões baseadas nos países...")

    # Criar mapeamento de países para sub-regiões
    sub_region_map = {
        # América do Norte
        'United States': '🌎 América do Norte', 'USA': '🌎 América do Norte',
        'Canada': '🌎 América do Norte', 'Mexico': '🌎 América do Norte',
        # América Central
        'Guatemala': '🌏 América Central', 'Belize': '🌏 América Central',
        'Honduras': '🌏 América Central', 'El Salvador': '🌏 América Central',
        'Nicaragua': '🌏 América Central', 'Costa Rica': '🌏 América Central',
        'Panama': '🌏 América Central',
        # Caribe
        'Cuba': '🏝️ Caribe', 'Jamaica': '🏝️ Caribe', 'Haiti': '🏝️ Caribe',
        'Dominican Republic': '🏝️ Caribe', 'Puerto Rico': '🏝️ Caribe',
        # América do Sul
        'Brazil': '🗺️ América do Sul', 'Brasil': '🗺️ América do Sul',
        'Argentina': '🗺️ América do Sul', 'Chile': '🗺️ América do Sul',
        'Peru': '🗺️ América do Sul', 'Colombia': '🗺️ América do Sul',
        'Venezuela': '🗺️ América do Sul', 'Ecuador': '🗺️ América do Sul',
        'Bolivia': '🗺️ América do Sul', 'Paraguay': '🗺️ América do Sul',
        'Uruguay': '🗺️ América do Sul'
    }

    df['sub_region'] = df['country'].map(sub_region_map)
    df['sub_region'] = df['sub_region'].fillna('📍 Outros')

# ===================================================================
# CÁLCULO DO PERFIL POR SUB-REGIÃO
# ===================================================================

# Calcular médias por sub-região
perfil_regiao = (df.groupby('sub_region')
                 .agg({
                     'danceability': 'mean',
                     'energy': 'mean',
                     'valence': 'mean',
                     'streams_millions': 'sum',
                     'track_name': 'count'
                 })
                 .round(3)
                 .reset_index())

# Renomear colunas
perfil_regiao.columns = ['sub_region', 'danceability', 'energy', 'valence', 'streams_millions', 'total_musicas']

# Ordenar por streams
perfil_regiao = perfil_regiao.sort_values('streams_millions', ascending=False)

# ===================================================================
# EXIBIÇÃO DO PERFIL
# ===================================================================

print("\n📊 PERFIL MUSICAL POR SUB-REGIÃO DAS AMÉRICAS:")
print("-" * 80)
print(f"{'Sub-região':22} | {'Streams(M)':>10} | {'Dança':>5} | {'Energia':>5} | {'Valence':>5} | {'Músicas':>8}")
print("-" * 80)

for _, row in perfil_regiao.iterrows():
    print(f"{row['sub_region']:22} | {row['streams_millions']:8.1f}M | {row['danceability']:5.2f} | {row['energy']:5.2f} | {row['valence']:5.2f} | {row['total_musicas']:8,}")

# ===================================================================
# GRÁFICO DE RADAR - COMPARAÇÃO ENTRE SUB-REGIÕES
# ===================================================================

# Selecionar top regiões para o radar (evitar muitas linhas)
top_regioes_radar = perfil_regiao.head(6)  # Máximo 6 para visualização limpa

# Cores para as regiões
cores_radar = {
    '🌎 América do Norte': '#4ECDC4',
    '🗺️ América do Sul': '#45B7D1',
    '🌏 América Central': '#FFE66D',
    '🏝️ Caribe': '#FF6B6B',
    '📍 Outros': '#95A5A6'
}

fig_radar = go.Figure()

for _, row in top_regioes_radar.iterrows():
    cor = cores_radar.get(row['sub_region'], '#96CEB4')
    fig_radar.add_trace(go.Scatterpolar(
        r=[row['danceability'], row['energy'], row['valence']],
        theta=['Dançabilidade', 'Energia', 'Positividade'],
        fill='toself',
        name=row['sub_region'],
        line=dict(color=cor),
        fillcolor=cor.replace('0.8', '0.2') if '#' in cor else cor
    ))

fig_radar.update_layout(
    title='🎵 Perfil Musical por Sub-região das Américas',
    polar=dict(
        radialaxis=dict(
            visible=True,
            range=[0, 1],
            tickvals=[0, 0.2, 0.4, 0.6, 0.8, 1]
        )),
    showlegend=True,
    height=600,
    title_x=0.5,
    title_font_size=18,
    legend=dict(
        x=1.05,
        y=1,
        bgcolor='rgba(255, 255, 255, 0.8)'
    )
)

fig_radar.show()
print("✅ Gráfico 1: Radar de perfil musical gerado!")

# ===================================================================
# GRÁFICO DE BARRAS AGRUPADAS
# ===================================================================

# Preparar dados para barras agrupadas
perfil_barras = perfil_regiao.melt(
    id_vars=['sub_region'],
    value_vars=['danceability', 'energy', 'valence'],
    var_name='caracteristica',
    value_name='valor'
)

# Mapeamento de nomes amigáveis
nome_caracteristicas = {
    'danceability': '💃 Dançabilidade',
    'energy': '⚡ Energia',
    'valence': '😊 Positividade'
}
perfil_barras['caracteristica'] = perfil_barras['caracteristica'].map(nome_caracteristicas)

fig_perfil_barras = px.bar(
    perfil_barras,
    x='sub_region',
    y='valor',
    color='caracteristica',
    title='🎸 Características Musicais por Sub-região das Américas',
    labels={'valor': 'Média (0-1)', 'sub_region': 'Sub-região', 'caracteristica': 'Característica'},
    barmode='group',
    color_discrete_map={
        '💃 Dançabilidade': '#FF6B6B',
        '⚡ Energia': '#4ECDC4',
        '😊 Positividade': '#45B7D1'
    }
)

fig_perfil_barras.update_layout(
    height=500,
    title_x=0.5,
    title_font_size=18,
    yaxis_range=[0, 1]
)

fig_perfil_barras.show()
print("✅ Gráfico 2: Barras comparativas gerado!")

# ===================================================================
# ANÁLISE DE DESTAQUE POR SUB-REGIÃO
# ===================================================================

print("\n" + "="*70)
print("🏆 DESTAQUES POR SUB-REGIÃO")
print("="*70)

# Encontrar maior valor para cada característica
caracteristicas = ['danceability', 'energy', 'valence']
nomes_carac = {'danceability': '💃 Dançabilidade', 'energy': '⚡ Energia', 'valence': '😊 Positividade'}

for carac in caracteristicas:
    max_regiao = perfil_regiao.loc[perfil_regiao[carac].idxmax()]
    min_regiao = perfil_regiao.loc[perfil_regiao[carac].idxmin()]
    print(f"\n{nomes_carac[carac]}:")
    print(f"   🏆 Maior: {max_regiao['sub_region']} ({max_regiao[carac]:.3f})")
    print(f"   📉 Menor: {min_regiao['sub_region']} ({min_regiao[carac]:.3f})")

# ===================================================================
# MAPA DE CALOR - COMPARAÇÃO ENTRE SUB-REGIÕES
# ===================================================================

# Criar matriz para heatmap
heatmap_data = perfil_regiao.set_index('sub_region')[['danceability', 'energy', 'valence']]
heatmap_data.columns = ['💃 Dançabilidade', '⚡ Energia', '😊 Positividade']

fig_heatmap = px.imshow(
    heatmap_data,
    text_auto=True,
    aspect='auto',
    title='🌡️ Mapa de Calor - Perfil Musical por Sub-região',
    labels=dict(x='Característica', y='Sub-região', color='Média (0-1)'),
    color_continuous_scale='Viridis',
    zmin=0,
    zmax=1
)

fig_heatmap.update_layout(
    height=450,
    title_x=0.5,
    title_font_size=16
)

fig_heatmap.show()
print("✅ Gráfico 3: Mapa de calor gerado!")

# ===================================================================
# INSIGHTS DO PERFIL MUSICAL
# ===================================================================

print("\n" + "="*70)
print("💡 INSIGHTS DO PERFIL MUSICAL DAS AMÉRICAS")
print("="*70)

# Encontrar região mais dançante
regiao_dancante = perfil_regiao.loc[perfil_regiao['danceability'].idxmax()]
regiao_energetica = perfil_regiao.loc[perfil_regiao['energy'].idxmax()]
regiao_feliz = perfil_regiao.loc[perfil_regiao['valence'].idxmax()]
regiao_total = perfil_regiao.loc[perfil_regiao['streams_millions'].idxmax()]

print(f"""
📊 PRINCIPAIS DESCOBERTAS:

   💃 MAIS DANÇANTE: {regiao_dancante['sub_region']} ({regiao_dancante['danceability']:.3f})
   ⚡ MAIS ENERGÉTICA: {regiao_energetica['sub_region']} ({regiao_energetica['energy']:.3f})
   😊 MAIS POSITIVA: {regiao_feliz['sub_region']} ({regiao_feliz['valence']:.3f})
   🎧 MAIS STREAMS: {regiao_total['sub_region']} ({regiao_total['streams_millions']:.1f}M streams)

📈 DIFERENÇAS ENTRE SUB-REGIÕES:
   • Danceability: {(perfil_regiao['danceability'].max() - perfil_regiao['danceability'].min()):.3f} de variação
   • Energia: {(perfil_regiao['energy'].max() - perfil_regiao['energy'].min()):.3f} de variação
   • Positividade: {(perfil_regiao['valence'].max() - perfil_regiao['valence'].min()):.3f} de variação
""")

print("✅ Análise de perfil musical das Américas concluída!")

# ===================================================================
# RANKING DE CARACTERÍSTICAS POR SUB-REGIÃO
# ===================================================================

print("\n🏆 RANKING POR CARACTERÍSTICA")
print("="*50)

for carac, nome in [('danceability', 'Dançabilidade'), ('energy', 'Energia'), ('valence', 'Positividade')]:
    print(f"\n{nome}:")
    ranking = perfil_regiao.sort_values(carac, ascending=False)[['sub_region', carac]].head(5)
    for i, (_, row) in enumerate(ranking.iterrows(), 1):
        print(f"   {i}. {row['sub_region']}: {row[carac]:.3f}")
```

```python
# ===================================================================
# ANÁLISE TEMPORAL - AMÉRICAS
# ===================================================================

print("📅 ANÁLISE TEMPORAL DAS AMÉRICAS")
print("=" * 70)

# Verificar se há dados de data
if 'week' not in df.columns:
    print("❌ Coluna 'week' não encontrada! Não é possível fazer análise temporal.")
    raise ValueError("Coluna week não disponível")

# Converter week para datetime se necessário
if not pd.api.types.is_datetime64_any_dtype(df['week']):
    df['week'] = pd.to_datetime(df['week'], errors='coerce')

# ===================================================================
# EVOLUÇÃO TEMPORAL POR SUB-REGIÃO (TOP 5)
# ===================================================================

# Escolher top 5 sub-regiões ou países
if 'sub_region' in df.columns:
    print("\n📊 Analisando evolução por SUB-REGIÃO...")
    top_regioes = df.groupby('sub_region')['streams_millions'].sum().nlargest(5).index
    df_temporal = df[df['sub_region'].isin(top_regioes)]
    coluna_agrupamento = 'sub_region'
    titulo_agrupamento = 'Sub-região'
    nome_legenda = 'Sub-região'
else:
    print("\n📊 Analisando evolução por PAÍS...")
    top5_paises = df.groupby('country')['streams_millions'].sum().nlargest(5).index
    df_temporal = df[df['country'].isin(top5_paises)]
    coluna_agrupamento = 'country'
    titulo_agrupamento = 'País'
    nome_legenda = 'País'

# Agregar dados temporalmente
temporal_semanal = df_temporal.groupby(['week', coluna_agrupamento])['streams_millions'].sum().reset_index()

print(f"\n📈 Analisando evolução de {len(temporal_semanal[coluna_agrupamento].unique())} {titulo_agrupamento}s principais")
print(f"{titulo_agrupamento}s: {', '.join(temporal_semanal[coluna_agrupamento].unique()[:5])}")

# ===================================================================
# GRÁFICO DE LINHA TEMPORAL
# ===================================================================

# Cores personalizadas para as Américas
cores_temporal = {
    '🌎 América do Norte': '#4ECDC4',
    '🗺️ América do Sul': '#45B7D1',
    '🌏 América Central': '#FFE66D',
    '🏝️ Caribe': '#FF6B6B',
    'Brazil': '#FF9F1C',
    'United States': '#2EC4B6',
    'Argentina': '#9B5DE5',
    'Mexico': '#FFB703'
}

fig_temporal = px.line(
    temporal_semanal,
    x='week',
    y='streams_millions',
    color=coluna_agrupamento,
    title=f'📈 Evolução Semanal de Streams - Top 5 {titulo_agrupamento}s das Américas',
    labels={'streams_millions': 'Streams (Milhões)', 'week': 'Semana', coluna_agrupamento: nome_legenda},
    line_shape='spline',
    color_discrete_map=cores_temporal if 'sub_region' in df.columns else None
)

fig_temporal.update_layout(
    height=550,
    hovermode='x unified',
    title_x=0.5,
    title_font_size=18,
    legend=dict(
        x=1.02,
        y=1,
        bgcolor='rgba(255, 255, 255, 0.8)'
    )
)

# Melhorar o formato do hover
fig_temporal.update_traces(
    hovertemplate='<b>%{fullData.name}</b><br>' +
                  'Semana: %{x|%d/%m/%Y}<br>' +
                  'Streams: %{y:.1f}M<br>' +
                  '<extra></extra>'
)

fig_temporal.show()
print("✅ Gráfico 1: Evolução temporal gerado!")

# ===================================================================
# ANÁLISE POR MÊS/ANO
# ===================================================================

print("\n📊 Analisando sazonalidade mensal...")

# Extrair componentes de data
df['year'] = df['week'].dt.year
df['month'] = df['week'].dt.month
df['month_name'] = df['week'].dt.strftime('%B')

# Agregar por mês
streams_mensal = df.groupby(['month_name', 'year'])['streams_millions'].sum().reset_index()

# Ordem correta dos meses
meses_ordem = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']
streams_mensal['month_name'] = pd.Categorical(streams_mensal['month_name'], categories=meses_ordem, ordered=True)
streams_mensal = streams_mensal.sort_values(['year', 'month_name'])

# Gráfico de barras mensal
fig_mensal = px.bar(
    streams_mensal,
    x='month_name',
    y='streams_millions',
    color='year',
    title='📊 Streams por Mês e Ano nas Américas',
    labels={'streams_millions': 'Streams (Milhões)', 'month_name': 'Mês', 'year': 'Ano'},
    barmode='group',
    color_discrete_sequence=px.colors.qualitative.Set2,
    text='streams_millions'
)

fig_mensal.update_traces(texttemplate='%{text:.1f}M', textposition='outside')
fig_mensal.update_layout(
    height=500,
    title_x=0.5,
    title_font_size=18
)
fig_mensal.show()
print("✅ Gráfico 2: Análise mensal gerada!")

# ===================================================================
# ANÁLISE POR DIA DA SEMANA
# ===================================================================

print("\n📊 Analisando padrões por dia da semana...")

# Extrair dia da semana
df['day_of_week'] = df['week'].dt.dayofweek
df['day_name'] = df['week'].dt.day_name()

# Ordem correta dos dias
dias_ordem = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dias_ordem_pt = ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta', 'Sábado', 'Domingo']

# Agregar por dia da semana
streams_diario = df.groupby('day_name')['streams_millions'].mean().reset_index()
streams_diario['day_name'] = pd.Categorical(streams_diario['day_name'], categories=dias_ordem, ordered=True)
streams_diario = streams_diario.sort_values('day_name')

# Gráfico de barras por dia
fig_diario = px.bar(
    streams_diario,
    x='day_name',
    y='streams_millions',
    title='📅 Média de Streams por Dia da Semana nas Américas',
    labels={'streams_millions': 'Média de Streams (Milhões)', 'day_name': 'Dia da Semana'},
    color='streams_millions',
    color_continuous_scale='Viridis',
    text='streams_millions'
)

fig_diario.update_traces(texttemplate='%{text:.1f}M', textposition='outside')
fig_diario.update_layout(height=450, title_x=0.5)
fig_diario.show()
print("✅ Gráfico 3: Análise por dia da semana gerada!")

# ===================================================================
# COMPARAÇÃO ENTRE PERÍODOS (se houver dados de 2 anos)
# ===================================================================

if len(df['year'].unique()) >= 2:
    print("\n📊 Comparando períodos...")

    # Agregar por ano e mês
    comparacao_anual = df.groupby(['year', 'month'])['streams_millions'].sum().reset_index()

    fig_comparacao = px.line(
        comparacao_anual,
        x='month',
        y='streams_millions',
        color='year',
        title='📈 Comparação Anual de Streams nas Américas',
        labels={'streams_millions': 'Streams (Milhões)', 'month': 'Mês', 'year': 'Ano'},
        markers=True
    )

    fig_comparacao.update_layout(height=450, title_x=0.5)
    fig_comparacao.update_xaxes(tickmode='array', tickvals=list(range(1, 13)), ticktext=meses_ordem_pt)
    fig_comparacao.show()
    print("✅ Gráfico 4: Comparação anual gerada!")

# ===================================================================
# INSIGHTS TEMPORAIS
# ===================================================================

print("\n" + "="*70)
print("💡 INSIGHTS DA ANÁLISE TEMPORAL")
print("="*70)

# Calcular métricas
if len(temporal_semanal) > 0:
    media_semanal = temporal_semanal.groupby(coluna_agrupamento)['streams_millions'].mean()
    max_semanal = temporal_semanal.loc[temporal_semanal.groupby(coluna_agrupamento)['streams_millions'].idxmax()]

    print(f"\n📊 MÉDIA SEMANAL DE STREAMS:")
    for item, media in media_semanal.sort_values(ascending=False).head(5).items():
        print(f"   • {item}: {media:.1f}M streams/semana")

    # Descobrir picos
    if len(max_semanal) > 0:
        print(f"\n🏆 RECORDES SEMANAIS:")
        for _, row in max_semanal.head(3).iterrows():
            print(f"   • {row[coluna_agrupamento]}: {row['streams_millions']:.1f}M (semana de {row['week'].date()})")

# Análise de sazonalidade mensal
mes_com_mais_streams = streams_mensal.loc[streams_mensal['streams_millions'].idxmax()]
mes_com_menos_streams = streams_mensal.loc[streams_mensal['streams_millions'].idxmin()]

print(f"\n📅 SAZONALIDADE:")
print(f"   • Mês com mais streams: {mes_com_mais_streams['month_name']} ({mes_com_mais_streams['year']}) - {mes_com_mais_streams['streams_millions']:.1f}M")
print(f"   • Mês com menos streams: {mes_com_menos_streams['month_name']} ({mes_com_menos_streams['year']}) - {mes_com_menos_streams['streams_millions']:.1f}M")

# Dia da semana com mais streams
dia_top = streams_diario.loc[streams_diario['streams_millions'].idxmax()]
print(f"\n📆 PADRÃO SEMANAL:")
print(f"   • Dia com mais streams: {dia_top['day_name']} (média de {dia_top['streams_millions']:.1f}M)")

print("\n✅ Análise temporal das Américas concluída!")

# ===================================================================
# MAPA DE CALOR TEMPORAL (MÊS vs DIA DA SEMANA)
# ===================================================================

print("\n📊 Mapa de Calor: Streams por Mês e Dia da Semana")

# Criar tabela pivô
df['month_num'] = df['week'].dt.month
heatmap_temporal = df.groupby(['month_num', 'day_of_week'])['streams_millions'].mean().unstack()

fig_heatmap_temporal = px.imshow(
    heatmap_temporal,
    text_auto='.1f',
    aspect='auto',
    title='🌡️ Média de Streams por Mês e Dia da Semana',
    labels=dict(x='Dia da Semana', y='Mês', color='Streams (M)'),
    color_continuous_scale='YlOrRd'
)

fig_heatmap_temporal.update_xaxis(ticktext=dias_ordem_pt, tickvals=list(range(7)))
fig_heatmap_temporal.update_yaxis(ticktext=meses_ordem_pt, tickvals=list(range(1, 13)))
fig_heatmap_temporal.update_layout(height=500, title_x=0.5)
fig_heatmap_temporal.show()

print("✅ Mapa de calor temporal gerado!")
```

```python
# ===================================================================
# ANÁLISE DE SUCESSO E CORRELAÇÕES - AMÉRICAS
# ===================================================================

print("🔥 ANÁLISE DE SUCESSO E CORRELAÇÕES NAS AMÉRICAS")
print("=" * 70)

# Verificar colunas disponíveis
colunas_disponiveis = df.columns.tolist()
print(f"\n📋 Colunas disponíveis para correlação: {[c for c in ['rank', 'streams_millions', 'danceability', 'energy', 'valence', 'success_score'] if c in colunas_disponiveis]}")

# ===================================================================
# MATRIZ DE CORRELAÇÃO
# ===================================================================

# Selecionar colunas para correlação
colunas_correlacao = []
for col in ['rank', 'streams_millions', 'danceability', 'energy', 'valence']:
    if col in df.columns:
        colunas_correlacao.append(col)

if 'success_score' in df.columns:
    colunas_correlacao.append('success_score')

if 'is_latin_style' in df.columns:
    colunas_correlacao.append('is_latin_style')

# Calcular matriz de correlação
corr_matrix = df[colunas_correlacao].corr()

# Renomear para nomes mais amigáveis
nome_amigavel = {
    'rank': 'Rank (posição)',
    'streams_millions': 'Streams (M)',
    'danceability': 'Dançabilidade',
    'energy': 'Energia',
    'valence': 'Positividade',
    'success_score': 'Score de Sucesso',
    'is_latin_style': 'Estilo Latino'
}

corr_matrix = corr_matrix.rename(index=nome_amigavel, columns=nome_amigavel)

# Heatmap de correlação
fig_heatmap = px.imshow(
    corr_matrix,
    text_auto=True,
    aspect='auto',
    title='📊 Matriz de Correlação - O que faz uma música ter sucesso nas Américas?',
    labels=dict(color='Correlação'),
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1
)

fig_heatmap.update_layout(
    height=600,
    width=700,
    title_x=0.5,
    title_font_size=18
)
fig_heatmap.show()
print("✅ Gráfico 1: Matriz de correlação gerada!")

# ===================================================================
# INSIGHTS DAS CORRELAÇÕES
# ===================================================================

print("\n📊 PRINCIPAIS CORRELAÇÕES ENCONTRADAS:")
print("-" * 50)

# Extrair correlações mais fortes com streams
if 'streams_millions' in corr_matrix.columns:
    streams_corr = corr_matrix['Streams (M)'].drop('Streams (M)').sort_values(ascending=False)

    for var, corr_val in streams_corr.items():
        if abs(corr_val) > 0.1:  # Mostrar apenas correlações relevantes
            direcao = "positiva" if corr_val > 0 else "negativa"
            intensidade = "forte" if abs(corr_val) > 0.5 else "moderada" if abs(corr_val) > 0.3 else "fraca"
            print(f"   • {var}: {direcao} {intensidade} ({corr_val:.3f})")

# ===================================================================
# SCATTER PLOT: DANÇABILIDADE VS ENERGIA
# ===================================================================

print("\n💃 Criando scatter plot: Dançabilidade vs Energia...")

# Amostragem segura
sample_size = min(3000, len(df))
df_sample = df.sample(sample_size, random_state=42)

# Adicionar sub-região para cor (se disponível)
if 'sub_region' in df.columns:
    fig_scatter = px.scatter(
        df_sample,
        x='danceability',
        y='energy',
        size='streams_millions',
        color='sub_region',
        hover_data=['track_name', 'artist_names', 'country', 'streams_millions'],
        title='💃 Dançabilidade vs Energia nas Américas (Tamanho = Streams)',
        labels={'danceability': 'Dançabilidade (0-1)', 'energy': 'Energia (0-1)', 'sub_region': 'Sub-região'},
        color_discrete_map={
            '🌎 América do Norte': '#4ECDC4',
            '🗺️ América do Sul': '#45B7D1',
            '🌏 América Central': '#FFE66D',
            '🏝️ Caribe': '#FF6B6B'
        },
        size_max=25,
        opacity=0.7
    )
else:
    fig_scatter = px.scatter(
        df_sample,
        x='danceability',
        y='energy',
        size='streams_millions',
        color='rank',
        hover_data=['track_name', 'artist_names', 'country', 'streams_millions'],
        title='💃 Dançabilidade vs Energia nas Américas (Tamanho = Streams)',
        labels={'danceability': 'Dançabilidade (0-1)', 'energy': 'Energia (0-1)', 'rank': 'Rank'},
        color_continuous_scale='Viridis_r',
        size_max=25,
        opacity=0.7
    )

fig_scatter.update_layout(
    height=600,
    title_x=0.5,
    title_font_size=18,
    xaxis_range=[0, 1],
    yaxis_range=[0, 1]
)

# Adicionar linha de referência (média)
fig_scatter.add_hline(y=df['energy'].mean(), line_dash="dash", line_color="gray", opacity=0.5)
fig_scatter.add_vline(x=df['danceability'].mean(), line_dash="dash", line_color="gray", opacity=0.5)

fig_scatter.show()
print("✅ Gráfico 2: Scatter plot gerado!")

# ===================================================================
# ANÁLISE DE SUCESSO POR RANK (BOXPLOT)
# ===================================================================

print("\n📊 Analisando distribuição de streams por grupo de rank...")

# Criar grupo de rank se não existir
if 'rank_group' not in df.columns:
    df['rank_group'] = pd.cut(
        df['rank'],
        bins=[0, 10, 50, 100, 200],
        labels=['🏆 Top 10', '📊 Top 11-50', '📈 Top 51-100', '📉 Top 101-200']
    )

# Boxplot por grupo de rank
boxplot_rank = px.box(
    df,
    x='rank_group',
    y='streams_millions',
    color='rank_group',
    title='📊 Distribuição de Streams por Posição no Ranking - Américas',
    labels={'rank_group': 'Posição no Ranking', 'streams_millions': 'Streams (Milhões)'},
    color_discrete_sequence=px.colors.qualitative.Set3
)

boxplot_rank.update_layout(
    height=500,
    title_x=0.5,
    title_font_size=18,
    showlegend=False
)

boxplot_rank.update_yaxes(title_text="Streams (Milhões)")
boxplot_rank.update_xaxes(title_text="")

boxplot_rank.show()
print("✅ Gráfico 3: Boxplot por rank gerado!")

# ===================================================================
# ANÁLISE DE SUCESSO POR SUB-REGIÃO
# ===================================================================

if 'sub_region' in df.columns:
    print("\n🏆 Comparando sucesso por sub-região...")

    # Calcular média de streams por sub-região
    streams_por_regiao = df.groupby('sub_region').agg({
        'streams_millions': ['mean', 'median', 'sum'],
        'rank': 'mean',
        'track_name': 'count'
    }).round(2)

    streams_por_regiao.columns = ['media_streams', 'mediana_streams', 'total_streams', 'rank_medio', 'total_musicas']
    streams_por_regiao = streams_por_regiao.sort_values('media_streams', ascending=False)

    print("\n📊 Média de Streams por Sub-região:")
    print("-" * 50)
    for regiao, row in streams_por_regiao.iterrows():
        print(f"   {regiao:22} : {row['media_streams']:.2f}M streams (rank médio: {row['rank_medio']:.1f})")

    # Gráfico de barras comparativo
    fig_comparacao_regioes = px.bar(
        streams_por_regiao.reset_index(),
        x='sub_region',
        y='media_streams',
        color='sub_region',
        text='media_streams',
        title='🎵 Média de Streams por Sub-região nas Américas',
        labels={'media_streams': 'Média de Streams (Milhões)', 'sub_region': 'Sub-região'},
        color_discrete_map={
            '🌎 América do Norte': '#4ECDC4',
            '🗺️ América do Sul': '#45B7D1',
            '🌏 América Central': '#FFE66D',
            '🏝️ Caribe': '#FF6B6B'
        }
    )

    fig_comparacao_regioes.update_traces(texttemplate='%{text:.2f}M', textposition='outside')
    fig_comparacao_regioes.update_layout(height=450, title_x=0.5, showlegend=False)
    fig_comparacao_regioes.show()
    print("✅ Gráfico 4: Comparação por sub-região gerado!")

# ===================================================================
# ANÁLISE DE SUCESSO POR ESTILO LATINO
# ===================================================================

if 'is_latin_style' in df.columns:
    print("\n🎵 Comparando músicas com estilo latino vs outras...")

    # Comparação
    latin_stats = df.groupby('is_latin_style').agg({
        'streams_millions': 'mean',
        'rank': 'mean',
        'danceability': 'mean',
        'energy': 'mean'
    }).round(3)

    latin_stats.index = ['Outros estilos', 'Estilo Latino']

    print("\n📊 Comparação de métricas:")
    print("-" * 60)
    print(latin_stats.to_string())

    # Gráfico de barras comparativo
    comparacao_latin = latin_stats.reset_index()
    comparacao_latin_melt = comparacao_latin.melt(id_vars='index', value_vars=['streams_millions', 'danceability', 'energy'])

    fig_latin = px.bar(
        comparacao_latin_melt,
        x='index',
        y='value',
        color='variable',
        title='🎤 Comparação: Estilo Latino vs Outros Estilos',
        labels={'value': 'Média', 'index': 'Tipo', 'variable': 'Métrica'},
        barmode='group',
        color_discrete_map={
            'streams_millions': '#FF6B6B',
            'danceability': '#4ECDC4',
            'energy': '#45B7D1'
        }
    )

    fig_latin.update_layout(height=450, title_x=0.5)
    fig_latin.show()
    print("✅ Gráfico 5: Comparação estilo latino gerado!")

# ===================================================================
# RESUMO DAS CORRELAÇÕES E SUCESSO
# ===================================================================

print("\n" + "="*70)
print("💡 RESUMO - O QUE FAZ UMA MÚSICA TER SUCESSO NAS AMÉRICAS?")
print("="*70)

# Encontrar a correlação mais forte com streams
if 'streams_millions' in corr_matrix.columns:
    corr_streams = corr_matrix['Streams (M)'].drop('Streams (M)')
    corr_streams = corr_streams[~corr_streams.index.isin(['Estilo Latino'])]

    if len(corr_streams) > 0:
        mais_correlacionado = corr_streams.abs().idxmax()
        valor_correlacao = corr_streams[mais_correlacionado]

        print(f"\n📈 FATOR MAIS CORRELACIONADO COM SUCESSO:")
        print(f"   • {mais_correlacionado}: {valor_correlacao:.3f}")

        if valor_correlacao > 0:
            print(f"   → Quanto MAIOR a {mais_correlacionado}, MAIOR os streams")
        else:
            print(f"   → Quanto MENOR a {mais_correlacionado}, MAIOR os streams")

# Média geral
print(f"\n📊 MÉDIAS GERAIS NAS AMÉRICAS:")
print(f"   • Média de streams: {df['streams_millions'].mean():.2f}M")
print(f"   • Dançabilidade média: {df['danceability'].mean():.3f}")
print(f"   • Energia média: {df['energy'].mean():.3f}")
print(f"   • Positividade média: {df['valence'].mean():.3f}")

print("\n✅ Análise de sucesso e correlações concluída!")

# ===================================================================
# CLUSTERIZAÇÃO DE ESTILOS MUSICAIS
# ===================================================================

print("\n🎯 Agrupamento de estilos musicais nas Américas")

# Gráfico 3D para visualização de clusters
fig_cluster = px.scatter_3d(
    df.sample(min(1000, len(df))),
    x='danceability',
    y='energy',
    z='valence',
    color='music_profile' if 'music_profile' in df.columns else None,
    size='streams_millions',
    hover_data=['track_name', 'artist_names', 'country'],
    title='🎨 Mapa 3D de Estilos Musicais - Américas',
    labels={'danceability': 'Dançabilidade', 'energy': 'Energia', 'valence': 'Positividade'}
)

fig_cluster.update_layout(height=700, title_x=0.5)
fig_cluster.show()
print("✅ Gráfico 3D de clusters gerado!")
```

```python
# ===================================================================
# DISTRIBUIÇÃO DE PERFIS MUSICAIS NAS AMÉRICAS
# ===================================================================

print("🎭 DISTRIBUIÇÃO DE PERFIS MUSICAIS NAS AMÉRICAS")
print("=" * 70)

# Verificar se a coluna music_profile existe
if 'music_profile' not in df.columns:
    print("⚠️ Coluna 'music_profile' não encontrada!")
    print("Criando perfis musicais baseados nas características...")

    def criar_perfil(row):
        if row['danceability'] > 0.7 and row['energy'] > 0.7:
            return '🔥 Ritmo Latino/Alta Energia'
        elif row['danceability'] > 0.7:
            return '💃 Dançante'
        elif row['energy'] > 0.7:
            return '⚡ Energética'
        elif row['valence'] > 0.7:
            return '😊 Feliz/Positiva'
        elif row['valence'] < 0.3:
            return '😢 Melancólica/Triste'
        else:
            return '🎵 Neutra/Equilibrada'

    df['music_profile'] = df.apply(criar_perfil, axis=1)

# ===================================================================
# CONTAGEM DE PERFIS
# ===================================================================

# Contagem de perfis
perfil_counts = df['music_profile'].value_counts().reset_index()
perfil_counts.columns = ['perfil', 'quantidade']

print("\n📊 DISTRIBUIÇÃO DOS PERFIS MUSICAIS NAS AMÉRICAS:")
print("-" * 70)
for _, row in perfil_counts.iterrows():
    percentual = (row['quantidade'] / len(df)) * 100
    barra = '█' * int(percentual / 2)
    print(f"{row['perfil']:28} : {row['quantidade']:8,} músicas ({percentual:5.1f}%) {barra}")

# ===================================================================
# GRÁFICO DE PIZZA (DONUT CHART)
# ===================================================================

# Cores personalizadas para os perfis
cores_perfis = {
    '🔥 Ritmo Latino/Alta Energia': '#FF6B6B',
    '💃 Dançante': '#4ECDC4',
    '⚡ Energética': '#45B7D1',
    '😊 Feliz/Positiva': '#96CEB4',
    '😢 Melancólica/Triste': '#D4A5A5',
    '🎵 Neutra/Equilibrada': '#95A5A6'
}

# Usar cores disponíveis no perfil
cores_disponiveis = [cores_perfis.get(p, '#95A5A6') for p in perfil_counts['perfil']]

fig_pizza = px.pie(
    perfil_counts,
    values='quantidade',
    names='perfil',
    title='🎵 Distribuição dos Perfis Musicais nas Américas',
    color_discrete_sequence=cores_disponiveis,
    hole=0.35
)

fig_pizza.update_traces(
    textposition='inside',
    textinfo='percent+label',
    pull=[0.05 if i == 0 else 0 for i in range(len(perfil_counts))]  # Destacar o maior
)
fig_pizza.update_layout(
    height=550,
    title_x=0.5,
    title_font_size=18,
    legend=dict(
        x=1.05,
        y=1,
        bgcolor='rgba(255, 255, 255, 0.8)'
    )
)
fig_pizza.show()
print("✅ Gráfico 1: Distribuição de perfis (pizza) gerado!")

# ===================================================================
# PERFIL MUSICAL POR SUB-REGIÃO
# ===================================================================

print("\n🎭 PERFIL MUSICAL POR SUB-REGIÃO DAS AMÉRICAS")
print("-" * 70)

# Usar sub-região se disponível, senão usar país
if 'sub_region' in df.columns:
    coluna_geografica = 'sub_region'
    titulo_geografico = 'Sub-região'
else:
    coluna_geografica = 'country'
    titulo_geografico = 'País'

# Calcular perfil por região
perfil_regiao = (df.groupby([coluna_geografica, 'music_profile'])
                 .size()
                 .reset_index(name='count'))

# Normalizar por região
perfil_regiao['percentual'] = perfil_regiao.groupby(coluna_geografica)['count'].transform(lambda x: x / x.sum() * 100)

# Selecionar top regiões (se muitas)
regioes_top = df.groupby(coluna_geografica)['streams_millions'].sum().nlargest(8).index
perfil_regiao_top = perfil_regiao[perfil_regiao[coluna_geografica].isin(regioes_top)]

# Gráfico de barras empilhadas
fig_perfil_regiao = px.bar(
    perfil_regiao_top,
    x=coluna_geografica,
    y='percentual',
    color='music_profile',
    title=f'🎸 Perfil Musical por {titulo_geografico} nas Américas (Top 8)',
    labels={'percentual': 'Percentual (%)', coluna_geografica: titulo_geografico, 'music_profile': 'Perfil Musical'},
    barmode='stack',
    color_discrete_map=cores_perfis
)

fig_perfil_regiao.update_layout(
    height=550,
    title_x=0.5,
    title_font_size=16,
    xaxis_title="",
    yaxis_title="Percentual (%)",
    legend=dict(
        x=1.02,
        y=1,
        bgcolor='rgba(255, 255, 255, 0.8)'
    )
)

fig_perfil_regiao.update_yaxis(range=[0, 100])
fig_perfil_regiao.show()
print("✅ Gráfico 2: Perfil por região gerado!")

# ===================================================================
# PERFIL MUSICAL POR SUB-REGIÃO (GRÁFICO DE ROSCA MÚLTIPLO)
# ===================================================================

if 'sub_region' in df.columns:
    print("\n📊 Comparação de perfis entre sub-regiões (gráfico de rosca múltiplo):")

    # Criar subplots para cada região
    regioes = df['sub_region'].unique()
    n_regioes = len(regioes)

    # Calcular dimensões do grid
    n_cols = min(2, n_regioes)
    n_rows = (n_regioes + n_cols - 1) // n_cols

    fig_donuts_multiplos = make_subplots(
        rows=n_rows, cols=n_cols,
        subplot_titles=regioes,
        specs=[[{'type': 'pie'} for _ in range(n_cols)] for _ in range(n_rows)]
    )

    row, col = 1, 1
    for regiao in regioes:
        # Dados da região
        dados_regiao = perfil_regiao[perfil_regiao[coluna_geografica] == regiao]

        if len(dados_regiao) > 0:
            fig_donuts_multiplos.add_trace(
                go.Pie(
                    labels=dados_regiao['music_profile'],
                    values=dados_regiao['count'],
                    hole=0.4,
                    showlegend=False,
                    textinfo='percent',
                    textposition='inside'
                ),
                row=row, col=col
            )

        col += 1
        if col > n_cols:
            col = 1
            row += 1

    fig_donuts_multiplos.update_layout(
        title_text="🍩 Perfil Musical por Sub-região das Américas",
        title_x=0.5,
        title_font_size=18,
        height=400 * n_rows,
        showlegend=False
    )

    fig_donuts_multiplos.show()
    print("✅ Gráfico 3: Múltiplos donuts gerado!")

# ===================================================================
# PERFIL MUSICAL POR PAÍS - MAPA DE CALOR
# ===================================================================

print("\n🌡️ Criando mapa de calor de perfis musicais...")

# Criar matriz de perfis por país (top 15 países)
top15_paises = df.groupby('country')['streams_millions'].sum().nlargest(15).index
df_top15 = df[df['country'].isin(top15_paises)]

# Matriz de contingência
matriz_perfis = pd.crosstab(df_top15['country'], df_top15['music_profile'], normalize='index') * 100

fig_heatmap_perfis = px.imshow(
    matriz_perfis,
    text_auto='.1f',
    aspect='auto',
    title='🌡️ Mapa de Calor - Perfil Musical por País (Top 15)',
    labels=dict(x='Perfil Musical', y='País', color='Percentual (%)'),
    color_continuous_scale='Viridis'
)

fig_heatmap_perfis.update_layout(
    height=550,
    width=800,
    title_x=0.5,
    title_font_size=16
)

fig_heatmap_perfis.show()
print("✅ Gráfico 4: Mapa de calor de perfis gerado!")

# ===================================================================
# INSIGHTS DOS PERFIS MUSICAIS
# ===================================================================

print("\n" + "="*70)
print("💡 INSIGHTS DOS PERFIS MUSICAIS NAS AMÉRICAS")
print("="*70)

# Perfil dominante geral
perfil_dominante = perfil_counts.iloc[0]['perfil']
pct_dominante = perfil_counts.iloc[0]['quantidade'] / len(df) * 100

print(f"\n🏆 PERFIL DOMINANTE GERAL:")
print(f"   • {perfil_dominante}: {pct_dominante:.1f}% das músicas")

# Perfil dominante por sub-região
if 'sub_region' in df.columns:
    print(f"\n🗺️ PERFIL DOMINANTE POR SUB-REGIÃO:")
    for regiao in df['sub_region'].unique():
        perfil_reg = df[df['sub_region'] == regiao]['music_profile'].mode()
        if len(perfil_reg) > 0:
            print(f"   • {regiao:20} : {perfil_reg.iloc[0]}")

# Perfis mais característicos (diferença em relação à média)
print(f"\n📊 PERFIS MAIS CARACTERÍSTICOS (destacam-se em relação à média):")
for regiao in df['sub_region'].unique()[:5]:
    df_regiao = df[df['sub_region'] == regiao]
    perfil_regiao_pct = df_regiao['music_profile'].value_counts(normalize=True) * 100
    perfil_global_pct = df['music_profile'].value_counts(normalize=True) * 100

    for perfil in perfil_regiao_pct.index[:2]:
        diferenca = perfil_regiao_pct[perfil] - (perfil_global_pct.get(perfil, 0))
        if diferenca > 5:
            print(f"   • {regiao}: +{diferenca:.1f}% em '{perfil}'")

print("\n✅ Análise de perfis musicais concluída!")

# ===================================================================
# TOP MÚSICAS POR PERFIL MUSICAL
# ===================================================================

print("\n🏆 TOP 3 MÚSICAS POR PERFIL MUSICAL")
print("="*60)

for perfil in df['music_profile'].unique():
    top_perfil = (df[df['music_profile'] == perfil]
                  .groupby(['track_name', 'artist_names'])['streams_millions']
                  .sum()
                  .sort_values(ascending=False)
                  .head(3))

    print(f"\n📌 {perfil}:")
    for i, (musica, streams) in enumerate(top_perfil.items(), 1):
        print(f"   {i}. {musica[0][:40]:40} : {streams:.1f}M streams")
```

```python
# ===================================================================
# ANÁLISE DE ARTISTAS E COLABORAÇÕES - AMÉRICAS
# ===================================================================

print("🎤 ANÁLISE DE ARTISTAS E COLABORAÇÕES NAS AMÉRICAS")
print("=" * 70)

# Verificar se há dados
if len(df) == 0:
    print("❌ DataFrame vazio! Carregue os dados primeiro.")
    raise ValueError("DataFrame vazio")

# ===================================================================
# TOP 10 ARTISTAS DAS AMÉRICAS
# ===================================================================

# Top 10 artistas
top_artistas = (df.groupby('artist_names')
                .agg({
                    'streams_millions': 'sum',
                    'track_name': 'nunique',
                    'country': 'nunique'
                })
                .rename(columns={'track_name': 'musicas_distintas', 'country': 'paises_presente'})
                .sort_values('streams_millions', ascending=False)
                .head(10)
                .reset_index())

# Adicionar sub-região principal do artista
if 'sub_region' in df.columns:
    artista_regiao = df.groupby('artist_names')['sub_region'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else 'N/A')
    top_artistas['sub_region'] = top_artistas['artist_names'].map(artista_regiao)

print("\n🏆 TOP 10 ARTISTAS MAIS OUVIDOS NAS AMÉRICAS:")
print("-" * 80)
for i, row in top_artistas.iterrows():
    regiao_info = f" [{row['sub_region']}]" if 'sub_region' in row else ""
    print(f"{i+1:2}. {row['artist_names'][:35]:35} : {row['streams_millions']:8.1f}M streams | {row['musicas_distintas']} músicas | {row['paises_presente']} países{regiao_info}")

# ===================================================================
# GRÁFICO DE BARRAS - TOP ARTISTAS
# ===================================================================

# Cores baseadas em sub-região (se disponível)
if 'sub_region' in top_artistas.columns:
    cores_artistas = {
        '🌎 América do Norte': '#4ECDC4',
        '🗺️ América do Sul': '#45B7D1',
        '🌏 América Central': '#FFE66D',
        '🏝️ Caribe': '#FF6B6B',
        'N/A': '#95A5A6'
    }

    fig_artistas = px.bar(
        top_artistas,
        x='streams_millions',
        y='artist_names',
        orientation='h',
        color='sub_region',
        text='streams_millions',
        title='🎤 Top 10 Artistas Mais Ouvidos nas Américas',
        labels={'streams_millions': 'Streams (Milhões)', 'artist_names': 'Artista', 'sub_region': 'Sub-região'},
        color_discrete_map=cores_artistas,
        hover_data=['musicas_distintas', 'paises_presente']
    )
else:
    fig_artistas = px.bar(
        top_artistas,
        x='streams_millions',
        y='artist_names',
        orientation='h',
        text='streams_millions',
        title='🎤 Top 10 Artistas Mais Ouvidos nas Américas',
        labels={'streams_millions': 'Streams (Milhões)', 'artist_names': 'Artista'},
        color='streams_millions',
        color_continuous_scale='Magma',
        hover_data=['musicas_distintas', 'paises_presente']
    )

fig_artistas.update_traces(texttemplate='%{text:.1f}M', textposition='outside')
fig_artistas.update_layout(
    height=550,
    title_x=0.5,
    title_font_size=18,
    xaxis_title="Streams (Milhões)",
    yaxis_title=""
)
fig_artistas.show()
print("✅ Gráfico 1: Top artistas gerado!")

# ===================================================================
# ANÁLISE DE COLABORAÇÕES
# ===================================================================

# Verificar se há coluna de colaboração
if 'collab' in df.columns:
    print("\n🤝 ANÁLISE DE COLABORAÇÕES NAS AMÉRICAS")
    print("-" * 50)

    # Converter collab para booleano se necessário
    if df['collab'].dtype == 'object':
        df['collab'] = df['collab'].map({'0': 0, '1': 1, 0: 0, 1: 1})

    # Estatísticas de colaborações
    collab_stats = df.groupby('collab').agg({
        'streams_millions': ['mean', 'median', 'count'],
        'track_name': 'nunique'
    })

    collab_stats.columns = ['media_streams', 'mediana_streams', 'total_registros', 'musicas_unicas']

    print(f"\n📊 ESTATÍSTICAS DE COLABORAÇÕES:")
    print("-" * 50)
    print(f"Músicas Solo:")
    print(f"   • Média de streams: {collab_stats.loc[0, 'media_streams']:.2f}M")
    print(f"   • Mediana de streams: {collab_stats.loc[0, 'mediana_streams']:.2f}M")
    print(f"   • Total de músicas: {collab_stats.loc[0, 'musicas_unicas']:,}")

    print(f"\nColaborações (2+ artistas):")
    print(f"   • Média de streams: {collab_stats.loc[1, 'media_streams']:.2f}M")
    print(f"   • Mediana de streams: {collab_stats.loc[1, 'mediana_streams']:.2f}M")
    print(f"   • Total de músicas: {collab_stats.loc[1, 'musicas_unicas']:,}")

    # Boxplot de colaborações
    fig_collab = px.box(
        df,
        x='collab',
        y='streams_millions',
        title='🤝 Colaborações vs Músicas Solo nas Américas',
        labels={'collab': 'Tipo', 'streams_millions': 'Streams (Milhões)'},
        color='collab',
        color_discrete_map={0: '#4ECDC4', 1: '#FF6B6B'}
    )

    fig_collab.update_layout(
        height=500,
        title_x=0.5,
        title_font_size=16
    )

    fig_collab.update_xaxes(ticktext=['Músicas Solo', 'Colaborações'], tickvals=[0, 1])
    fig_collab.show()
    print("✅ Gráfico 2: Análise de colaborações gerada!")

    # Top colaborações
    print(f"\n🎵 TOP COLABORAÇÕES NAS AMÉRICAS:")
    colaboracoes = df[df['collab'] == 1].groupby(['track_name', 'artist_names'])['streams_millions'].sum().sort_values(ascending=False).head(10)
    for i, (musica, streams) in enumerate(colaboracoes.items(), 1):
        print(f"   {i}. {musica[0][:40]:40} : {streams:.1f}M streams")

else:
    print("\n⚠️ Coluna 'collab' não encontrada!")
    print("   Pulando análise de colaborações.")

# ===================================================================
# ANÁLISE DE DOMÍNIO POR SUB-REGIÃO
# ===================================================================

print("\n🗺️ ANÁLISE DE DOMÍNIO POR SUB-REGIÃO")
print("-" * 50)

# Encontrar artistas mais populares por sub-região
if 'sub_region' in df.columns:
    for regiao in df['sub_region'].unique():
        df_regiao = df[df['sub_region'] == regiao]
        if len(df_regiao) > 0:
            top_artista_regiao = (df_regiao.groupby('artist_names')['streams_millions']
                                  .sum()
                                  .sort_values(ascending=False)
                                  .head(1))

            if len(top_artista_regiao) > 0:
                artista = top_artista_regiao.index[0]
                streams = top_artista_regiao.values[0]
                print(f"   {regiao:22} : {artista[:30]:30} ({streams:.1f}M streams)")

# ===================================================================
# DIVERSIDADE DE ARTISTAS POR PAÍS
# ===================================================================

print("\n🎨 DIVERSIDADE DE ARTISTAS POR PAÍS")
print("-" * 50)

# Top países com maior diversidade de artistas
diversidade_artistas = (df.groupby('country')
                        .agg({
                            'artist_names': 'nunique',
                            'streams_millions': 'sum'
                        })
                        .rename(columns={'artist_names': 'total_artistas'})
                        .sort_values('total_artistas', ascending=False)
                        .head(8)
                        .reset_index())

print("\n🏆 PAÍSES COM MAIS ARTISTAS ÚNICOS:")
print("-" * 60)
for _, row in diversidade_artistas.iterrows():
    print(f"   {row['country']:22} : {row['total_artistas']:5,} artistas | {row['streams_millions']:.1f}M streams")

# Gráfico de diversidade
fig_diversidade = px.bar(
    diversidade_artistas,
    x='country',
    y='total_artistas',
    color='streams_millions',
    text='total_artistas',
    title='🎨 Países com Maior Diversidade de Artistas nas Américas',
    labels={'country': 'País', 'total_artistas': 'Número de Artistas', 'streams_millions': 'Streams (M)'},
    color_continuous_scale='Teal'
)

fig_diversidade.update_traces(texttemplate='%{text:,}', textposition='outside')
fig_diversidade.update_layout(height=450, title_x=0.5)
fig_diversidade.show()
print("✅ Gráfico 3: Diversidade de artistas gerada!")

# ===================================================================
# ARTISTAS QUE MAIS APARECEM EM MÚLTIPLOS PAÍSES
# ===================================================================

print("\n🌍 ARTISTAS COM MAIOR ALCANCE GEOGRÁFICO")
print("-" * 50)

# Artistas que aparecem em mais países
alcance_artistas = (df.groupby('artist_names')['country']
                    .nunique()
                    .sort_values(ascending=False)
                    .head(10)
                    .reset_index())
alcance_artistas.columns = ['artist_names', 'paises_alcancados']

print("\n🏆 ARTISTAS MAIS INTERNACIONAIS:")
for _, row in alcance_artistas.iterrows():
    print(f"   • {row['artist_names'][:35]:35} : {row['paises_alcancados']} países")

# ===================================================================
# INSIGHTS DA ANÁLISE DE ARTISTAS
# ===================================================================

print("\n" + "="*70)
print("💡 INSIGHTS DA ANÁLISE DE ARTISTAS")
print("="*70)

print(f"""
🏆 PRINCIPAIS DESCOBERTAS:

   • Artista mais ouvido: {top_artistas.iloc[0]['artist_names']} ({top_artistas.iloc[0]['streams_millions']:.1f}M streams)
   • Artista com mais músicas distintas: {top_artistas.loc[top_artistas['musicas_distintas'].idxmax(), 'artist_names']} ({top_artistas['musicas_distintas'].max()} músicas)
   • Artista com maior alcance: {alcance_artistas.iloc[0]['artist_names']} ({alcance_artistas.iloc[0]['paises_alcancados']} países)
""")

if 'collab' in df.columns:
    diferenca = ((collab_stats.loc[1, 'media_streams'] - collab_stats.loc[0, 'media_streams']) / collab_stats.loc[0, 'media_streams']) * 100
    if diferenca > 0:
        print(f"   • Colaborações têm em média {diferenca:.1f}% mais streams que músicas solo")
    else:
        print(f"   • Músicas solo têm em média {abs(diferenca):.1f}% mais streams que colaborações")

print("\n✅ Análise de artistas e colaborações concluída!")

# ===================================================================
# REDE DE COLABORAÇÕES ENTRE ARTISTAS (BÔNUS)
# ===================================================================

print("\n🕸️ MAPA DE COLABORAÇÕES ENTRE ARTISTAS")

# Identificar colaborações (quando vários artistas aparecem juntos)
if 'artist_names' in df.columns:
    # Buscar músicas com múltiplos artistas (contém "feat" ou "&")
    colaboracoes = df[df['artist_names'].str.contains('feat|&|,', na=False)]

    if len(colaboracoes) > 0:
        # Contar colaborações mais frequentes
        top_colaboracoes = colaboracoes['artist_names'].value_counts().head(10)

        print("\n🎵 COLABORAÇÕES MAIS FREQUENTES:")
        for i, (colab, qtd) in enumerate(top_colaboracoes.items(), 1):
            print(f"   {i}. {colab[:50]:50} : {qtd} aparições")
    else:
        print("   Nenhuma colaboração identificada no formato 'feat' ou '&'")
```

```python
# ===================================================================
# DASHBOARD INTERATIVO INTEGRADO - AMÉRICAS
# ===================================================================

print("📊 CRIANDO DASHBOARD INTERATIVO DAS AMÉRICAS...")
print("=" * 70)

from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px

# ===================================================================
# PREPARAÇÃO DOS DADOS
# ===================================================================

# Cores personalizadas para as Américas
CORES_DASHBOARD = {
    'América do Norte': '#4ECDC4',
    'América do Sul': '#45B7D1',
    'América Central': '#FFE66D',
    'Caribe': '#FF6B6B',
    'default': '#96CEB4'
}

# Top 5 países (baseado em streams)
top5_paises = df.groupby('country')['streams_millions'].sum().nlargest(5).reset_index()

# Top 3 países para evolução temporal
top3_paises = df.groupby('country')['streams_millions'].sum().nlargest(3).index

# Dados temporais
df_top3 = df[df['country'].isin(top3_paises)]
temporal_top3 = df_top3.groupby(['week', 'country'])['streams_millions'].sum().reset_index()

# Perfil musical médio global
medias_globais = df[['danceability', 'energy', 'valence']].mean()

# Perfil musical por sub-região (se disponível)
if 'sub_region' in df.columns:
    medias_regioes = df.groupby('sub_region')[['danceability', 'energy', 'valence']].mean()
else:
    medias_regioes = None

# Verificar se rank_group existe
if 'rank_group' not in df.columns:
    df['rank_group'] = pd.cut(df['rank'], bins=[0, 10, 50, 100, 200],
                               labels=['🏆 Top 10', '📊 Top 11-50', '📈 Top 51-100', '📉 Top 101-200'])

# ===================================================================
# CRIAÇÃO DO DASHBOARD
# ===================================================================

# Criar dashboard com 2x2 grid
fig_dashboard = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        '🌍 Top 5 Países por Streams nas Américas',
        '🎵 Perfil Musical Médio - Américas',
        '📈 Evolução Temporal (Top 3 Países)',
        '📊 Distribuição de Streams por Ranking'
    ),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'scatter'}, {'type': 'box'}]],
    vertical_spacing=0.12,
    horizontal_spacing=0.1
)

# ===================================================================
# GRÁFICO 1: TOP 5 PAÍSES (BARRAS)
# ===================================================================

# Cores baseadas em sub-região se disponível
if 'sub_region' in df.columns:
    # Obter sub-região de cada país
    pais_regiao = df.groupby('country')['sub_region'].first().to_dict()
    cores_paises = [CORES_DASHBOARD.get(pais_regiao.get(p, 'default'), CORES_DASHBOARD['default'])
                    for p in top5_paises['country']]
else:
    cores_paises = px.colors.qualitative.Set3[:5]

fig_dashboard.add_trace(
    go.Bar(
        x=top5_paises['country'],
        y=top5_paises['streams_millions'],
        marker_color=cores_paises,
        text=top5_paises['streams_millions'].round(1),
        textposition='outside',
        name='Streams',
        hovertemplate='<b>%{x}</b><br>Streams: %{y:.1f}M<extra></extra>'
    ),
    row=1, col=1
)

# ===================================================================
# GRÁFICO 2: PERFIL MUSICAL MÉDIO (BARRAS AGRUPADAS POR SUB-REGIÃO OU GLOBAL)
# ===================================================================

if medias_regioes is not None and len(medias_regioes) <= 4:
    # Mostrar comparação entre sub-regiões
    for regiao in medias_regioes.index:
        fig_dashboard.add_trace(
            go.Bar(
                x=['Dançabilidade', 'Energia', 'Positividade'],
                y=medias_regioes.loc[regiao, ['danceability', 'energy', 'valence']],
                name=regiao,
                marker_color=CORES_DASHBOARD.get(regiao, CORES_DASHBOARD['default']),
                text=medias_regioes.loc[regiao, ['danceability', 'energy', 'valence']].round(2),
                textposition='inside'
            ),
            row=1, col=2
        )
else:
    # Mostrar média global
    fig_dashboard.add_trace(
        go.Bar(
            x=['💃 Dançabilidade', '⚡ Energia', '😊 Positividade'],
            y=medias_globais.values,
            marker_color=['#FF6B6B', '#4ECDC4', '#45B7D1'],
            text=medias_globais.values.round(2),
            textposition='auto',
            name='Média Global',
            hovertemplate='<b>%{x}</b><br>Média: %{y:.3f}<extra></extra>'
        ),
        row=1, col=2
    )

# ===================================================================
# GRÁFICO 3: EVOLUÇÃO TEMPORAL (LINHAS)
# ===================================================================

# Cores para os top 3 países
cores_temporal = {
    'United States': '#2EC4B6',
    'Brazil': '#FF9F1C',
    'Mexico': '#FFB703',
    'Argentina': '#9B5DE5',
    'Canada': '#4ECDC4'
}

for pais in top3_paises:
    dados_pais = temporal_top3[temporal_top3['country'] == pais]
    cor = cores_temporal.get(pais, '#96CEB4')

    fig_dashboard.add_trace(
        go.Scatter(
            x=dados_pais['week'],
            y=dados_pais['streams_millions'],
            name=pais,
            mode='lines',
            line=dict(width=2.5, color=cor),
            fill='tonexty' if pais == top3_paises[0] else None,
            opacity=0.8,
            hovertemplate='<b>%{fullData.name}</b><br>Semana: %{x|%d/%m/%Y}<br>Streams: %{y:.1f}M<extra></extra>'
        ),
        row=2, col=1
    )

# ===================================================================
# GRÁFICO 4: BOXPLOT POR GRUPO DE RANK
# ===================================================================

# Ordem correta dos grupos de rank
ordem_rank = ['🏆 Top 10', '📊 Top 11-50', '📈 Top 51-100', '📉 Top 101-200']

fig_dashboard.add_trace(
    go.Box(
        x=df['rank_group'],
        y=df['streams_millions'],
        marker_color='lightcoral',
        boxmean='sd',
        hovertemplate='<b>%{x}</b><br>Streams: %{y:.1f}M<br>Mediana: %{boxmean:.1f}M<extra></extra>'
    ),
    row=2, col=2
)

# ===================================================================
# ATUALIZAÇÃO DO LAYOUT
# ===================================================================

# Atualizar layout do dashboard
fig_dashboard.update_layout(
    title_text="🌎 DASHBOARD SPOTIFY - ANÁLISE DAS AMÉRICAS",
    title_x=0.5,
    title_font_size=22,
    height=850,
    width=1100,
    showlegend=True if medias_regioes is not None else False,
    legend=dict(
        x=1.02,
        y=0.98,
        bgcolor='rgba(255, 255, 255, 0.8)',
        bordercolor='lightgray',
        borderwidth=1
    ),
    plot_bgcolor='rgba(240, 240, 240, 0.5)',
    paper_bgcolor='white',
    font=dict(family="Arial, sans-serif", size=12)
)

# Configurar eixos
# Gráfico 1
fig_dashboard.update_xaxes(title_text="País", row=1, col=1, tickangle=0)
fig_dashboard.update_yaxes(title_text="Streams (Milhões)", row=1, col=1, gridcolor='lightgray')

# Gráfico 2
fig_dashboard.update_xaxes(title_text="Característica Musical", row=1, col=2)
fig_dashboard.update_yaxes(title_text="Média (0-1)", row=1, col=2, range=[0, 1], gridcolor='lightgray')

# Gráfico 3
fig_dashboard.update_xaxes(title_text="Semana", row=2, col=1, tickangle=45)
fig_dashboard.update_yaxes(title_text="Streams (Milhões)", row=2, col=1, gridcolor='lightgray')

# Gráfico 4
fig_dashboard.update_xaxes(title_text="Grupo de Ranking", row=2, col=2)
fig_dashboard.update_yaxes(title_text="Streams (Milhões)", row=2, col=2, gridcolor='lightgray')

# Adicionar anotações com métricas importantes
fig_dashboard.add_annotation(
    text=f"📊 Total de streams: {df['streams_millions'].sum():.0f}M | 🎵 Músicas: {df['track_name'].nunique():,} | 🌎 Países: {df['country'].nunique()}",
    xref="paper", yref="paper",
    x=0.5, y=1.02,
    showarrow=False,
    font=dict(size=11, color='gray')
)

fig_dashboard.show()
print("✅ Dashboard integrado das Américas gerado com sucesso!")

# ===================================================================
# DASHBOARD COMPLEMENTAR - VISÃO POR SUB-REGIÃO (OPCIONAL)
# ===================================================================

if 'sub_region' in df.columns:
    print("\n📊 CRIANDO DASHBOARD COMPLEMENTAR POR SUB-REGIÃO...")

    fig_dashboard_regioes = make_subplots(
        rows=2, cols=2,
        subplot_titles=(
            'Streams por Sub-região',
            'Perfil Musical por Sub-região',
            'Distribuição de Perfis Musicais',
            'Top Artistas por Sub-região'
        ),
        specs=[[{'type': 'bar'}, {'type': 'bar'}],
               [{'type': 'pie'}, {'type': 'bar'}]]
    )

    # Gráfico 1: Streams por sub-região
    streams_regiao = df.groupby('sub_region')['streams_millions'].sum().sort_values(ascending=False).reset_index()
    cores_regioes_lista = [CORES_DASHBOARD.get(r, CORES_DASHBOARD['default']) for r in streams_regiao['sub_region']]

    fig_dashboard_regioes.add_trace(
        go.Bar(
            x=streams_regiao['sub_region'],
            y=streams_regiao['streams_millions'],
            marker_color=cores_regioes_lista,
            text=streams_regiao['streams_millions'].round(1)
        ),
        row=1, col=1
    )

    # Gráfico 2: Perfil musical por sub-região
    for regiao in medias_regioes.index[:4]:
        fig_dashboard_regioes.add_trace(
            go.Bar(
                x=['Dança', 'Energia', 'Valence'],
                y=medias_regioes.loc[regiao],
                name=regiao,
                marker_color=CORES_DASHBOARD.get(regiao, CORES_DASHBOARD['default'])
            ),
            row=1, col=2
        )

    # Gráfico 3: Distribuição de perfis musicais (pizza)
    if 'music_profile' in df.columns:
        top_perfis = df['music_profile'].value_counts().head(4).reset_index()
        top_perfis.columns = ['perfil', 'quantidade']

        fig_dashboard_regioes.add_trace(
            go.Pie(
                labels=top_perfis['perfil'],
                values=top_perfis['quantidade'],
                hole=0.3,
                textinfo='percent+label'
            ),
            row=2, col=1
        )

    # Gráfico 4: Top artistas da sub-região principal
    principal_regiao = streams_regiao.iloc[0]['sub_region']
    top_artistas_regiao = (df[df['sub_region'] == principal_regiao]
                           .groupby('artist_names')['streams_millions']
                           .sum()
                           .sort_values(ascending=False)
                           .head(5)
                           .reset_index())

    fig_dashboard_regioes.add_trace(
        go.Bar(
            x=top_artistas_regiao['streams_millions'],
            y=top_artistas_regiao['artist_names'],
            orientation='h',
            marker_color='#45B7D1',
            text=top_artistas_regiao['streams_millions'].round(1)
        ),
        row=2, col=2
    )

    fig_dashboard_regioes.update_layout(
        title_text="🌎 DASHBOARD POR SUB-REGIÃO - AMÉRICAS",
        title_x=0.5,
        height=800,
        showlegend=True if medias_regioes is not None else False
    )

    fig_dashboard_regioes.show()
    print("✅ Dashboard por sub-região gerado com sucesso!")

# ===================================================================
# INSIGHTS DO DASHBOARD
# ===================================================================

print("\n" + "="*70)
print("💡 PRINCIPAIS INSIGHTS DO DASHBOARD")
print("="*70)

print(f"""
📊 MÉTRICAS PRINCIPAIS:

   • 🌍 País com mais streams: {top5_paises.iloc[0]['country']} ({top5_paises.iloc[0]['streams_millions']:.1f}M)
   • 🎵 Perfil musical dominante: {'Dançabilidade' if medias_globais['danceability'] > medias_globais['energy'] else 'Energia'}
   • 📈 Média geral de streams: {df['streams_millions'].mean():.2f}M por música
   • 🏆 Ranking médio: {df['rank'].mean():.1f}

🎯 RECOMENDAÇÕES:
   • Para lançamentos, foque nos {meses_com_mais_streams} meses com maior volume
   • Invista em playlists focadas no perfil {'dançante' if medias_globais['danceability'] > 0.65 else 'energético'}
   • Considere colaborações entre artistas de diferentes sub-regiões
""")

print("✅ Dashboard finalizado! Explore os gráficos interativos acima.")
print("="*70)
```

```python
# ===================================================================
# INSIGHTS E PRINCIPAIS DESCOBERTAS - AMÉRICAS
# ===================================================================

print("💡 INSIGHTS E PRINCIPAIS DESCOBERTAS DO MERCADO MUSICAL DAS AMÉRICAS")
print("=" * 80)

# ===================================================================
# MÉTRICAS GLOBAIS
# ===================================================================

total_streams = df['streams_millions'].sum()
media_dance = df['danceability'].mean()
media_energy = df['energy'].mean()
media_valence = df['valence'].mean()
media_streams = df['streams_millions'].mean()
mediana_streams = df['streams_millions'].median()

# Top música
top_musica = (df.groupby(['track_name', 'artist_names'])['streams_millions']
              .sum()
              .sort_values(ascending=False)
              .head(1))

# Top país
top_pais = (df.groupby('country')['streams_millions']
            .sum()
            .sort_values(ascending=False)
            .head(1))

# Top artista
top_artista = (df.groupby('artist_names')['streams_millions']
               .sum()
               .sort_values(ascending=False)
               .head(1))

# Correlações
corr_rank_streams = df['rank'].corr(df['streams_millions'])
corr_dance_streams = df['danceability'].corr(df['streams_millions'])
corr_energy_streams = df['energy'].corr(df['streams_millions'])

# Perfil dominante
if 'music_profile' in df.columns:
    perfil_dominante = df['music_profile'].mode()[0]
    pct_perfil_dominante = (df['music_profile'] == perfil_dominante).sum() / len(df) * 100
else:
    perfil_dominante = "N/A"
    pct_perfil_dominante = 0

# ===================================================================
# MÉTRICAS GLOBAIS
# ===================================================================

print("\n📊 MÉTRICAS GLOBAIS DAS AMÉRICAS:")
print("-" * 60)
print(f"🎵 Total de streams analisados: {total_streams:.1f}M")
print(f"📊 Média de streams por música: {media_streams:.2f}M")
print(f"📈 Mediana de streams: {mediana_streams:.2f}M")
print(f"🌍 Países representados: {df['country'].nunique()}")
print(f"🎤 Artistas únicos: {df['artist_names'].nunique():,}")
print(f"🎵 Músicas únicas: {df['track_name'].nunique():,}")

# ===================================================================
# TOP PERFORMERS
# ===================================================================

print("\n🏆 TOP PERFORMERS DAS AMÉRICAS:")
print("-" * 60)
print(f"🎵 Música mais ouvida: {top_musica.index[0][0][:50]}...")
print(f"   → {top_musica.iloc[0]['artist_names']} ({top_musica['streams_millions'].values[0]:.1f}M streams)")
print(f"🌍 País com mais streams: {top_pais.index[0]} ({top_pais.values[0]:.1f}M streams)")
print(f"🎤 Artista mais ouvido: {top_artista.index[0]} ({top_artista.values[0]:.1f}M streams)")

# ===================================================================
# CARACTERÍSTICAS MUSICAIS
# ===================================================================

print("\n🎸 CARACTERÍSTICAS MUSICAIS DAS AMÉRICAS:")
print("-" * 60)
print(f"💃 Dançabilidade média: {media_dance:.3f}")
print(f"⚡ Energia média: {media_energy:.3f}")
print(f"😊 Positividade (Valence) média: {media_valence:.3f}")

# Comparação com benchmarks
print(f"\n📊 ANÁLISE DAS CARACTERÍSTICAS:")
if media_dance > 0.65:
    print(f"   • Dançabilidade ALTA → Músicas nas Américas são muito dançantes")
elif media_dance > 0.55:
    print(f"   • Dançabilidade MÉDIA → Equilíbrio entre ritmo e melodia")
else:
    print(f"   • Dançabilidade BAIXA → Preferência por músicas mais calmas")

if media_valence > 0.55:
    print(f"   • Positividade ALTA → Preferência por músicas alegres/felizes")
elif media_valence > 0.45:
    print(f"   • Positividade MÉDIA → Variedade de emoções musicais")
else:
    print(f"   • Positividade BAIXA → Preferência por músicas mais melancólicas")

# ===================================================================
# CORRELAÇÕES IMPORTANTES
# ===================================================================

print("\n📈 CORRELAÇÕES IMPORTANTES (O que faz uma música ter sucesso?):")
print("-" * 60)
print(f"📊 Correlação Rank vs Streams: {corr_rank_streams:.3f}")
print(f"   → {'Forte relação' if abs(corr_rank_streams) > 0.5 else 'Relação moderada' if abs(corr_rank_streams) > 0.3 else 'Baixa relação'} (negativo = melhor rank = mais streams)")

print(f"\n📊 Correlação Dançabilidade vs Streams: {corr_dance_streams:.3f}")
if corr_dance_streams > 0.2:
    print(f"   → Músicas MAIS dançantes tendem a ter MAIS streams")
elif corr_dance_streams < -0.2:
    print(f"   → Músicas MENOS dançantes tendem a ter MAIS streams")
else:
    print(f"   → Dançabilidade tem pouca influência no sucesso")

print(f"\n📊 Correlação Energia vs Streams: {corr_energy_streams:.3f}")
if corr_energy_streams > 0.2:
    print(f"   → Músicas MAIS energéticas tendem a ter MAIS streams")
elif corr_energy_streams < -0.2:
    print(f"   → Músicas MENOS energéticas tendem a ter MAIS streams")
else:
    print(f"   → Energia tem pouca influência no sucesso")

# ===================================================================
# PERFIL DOMINANTE
# ===================================================================

print("\n🎭 PERFIL MUSICAL DOMINANTE NAS AMÉRICAS:")
print("-" * 60)
print(f"🎵 Tipo musical mais comum: {perfil_dominante}")
print(f"   → Representa {pct_perfil_dominante:.1f}% de todas as músicas")

# ===================================================================
# INSIGHTS ADICIONAIS (quando disponíveis)
# ===================================================================

# Análise por sub-região
if 'sub_region' in df.columns:
    print("\n🗺️ INSIGHTS POR SUB-REGIÃO:")
    print("-" * 60)

    for regiao in df['sub_region'].unique():
        df_reg = df[df['sub_region'] == regiao]
        pct_streams = (df_reg['streams_millions'].sum() / total_streams) * 100
        dance_reg = df_reg['danceability'].mean()

        print(f"   • {regiao}:")
        print(f"     → {pct_streams:.1f}% dos streams | Dançabilidade: {dance_reg:.3f}")

        # Artista mais ouvido da região
        top_artist_reg = (df_reg.groupby('artist_names')['streams_millions']
                          .sum()
                          .sort_values(ascending=False)
                          .head(1))
        if len(top_artist_reg) > 0:
            print(f"     → Artista destaque: {top_artist_reg.index[0][:35]}")

# Análise de colaborações
if 'collab' in df.columns:
    print("\n🤝 INSIGHTS SOBRE COLABORAÇÕES:")
    print("-" * 60)

    media_solo = df[df['collab'] == 0]['streams_millions'].mean()
    media_collab = df[df['collab'] == 1]['streams_millions'].mean()

    print(f"   • Média de streams (solo): {media_solo:.2f}M")
    print(f"   • Média de streams (colab): {media_collab:.2f}M")

    if media_collab > media_solo:
        diferenca = ((media_collab - media_solo) / media_solo) * 100
        print(f"   → Colaborações têm {diferenca:.1f}% MAIS streams que músicas solo")
    else:
        diferenca = ((media_solo - media_collab) / media_collab) * 100
        print(f"   → Músicas solo têm {diferenca:.1f}% MAIS streams que colaborações")

# Análise de estilo latino
if 'is_latin_style' in df.columns:
    print("\n🎵 INSIGHTS SOBRE ESTILO LATINO:")
    print("-" * 60)

    pct_latin = df['is_latin_style'].mean() * 100
    media_latin = df[df['is_latin_style'] == 1]['streams_millions'].mean()
    media_nao_latin = df[df['is_latin_style'] == 0]['streams_millions'].mean()

    print(f"   • Músicas com estilo latino: {pct_latin:.1f}% do total")
    print(f"   • Média de streams (latino): {media_latin:.2f}M")
    print(f"   • Média de streams (outros): {media_nao_latin:.2f}M")

    if media_latin > media_nao_latin:
        print(f"   → Estilo latino tem MAIOR sucesso nas Américas")
    else:
        print(f"   → Estilo latino tem MENOR sucesso nas Américas")

# ===================================================================
# RECOMENDAÇÕES ESTRATÉGICAS
# ===================================================================

print("\n" + "="*80)
print("🎯 RECOMENDAÇÕES ESTRATÉGICAS PARA O MERCADO DAS AMÉRICAS")
print("="*80)

recomendacoes = []

# Baseado nas correlações
if corr_dance_streams > 0.2:
    recomendacoes.append("💃 INVESTIR em músicas com ALTA DANÇABILIDADE (>0.7)")
elif corr_energy_streams > 0.2:
    recomendacoes.append("⚡ INVESTIR em músicas com ALTA ENERGIA (>0.7)")
else:
    recomendacoes.append("🎵 MANTER diversidade musical - características variadas têm sucesso")

# Baseado no perfil dominante
if perfil_dominante != "N/A":
    if "Dançante" in perfil_dominante or "Latino" in perfil_dominante:
        recomendacoes.append("🔥 FORTALECER presença em playlists de música latina e dançante")
    elif "Energética" in perfil_dominante:
        recomendacoes.append("⚡ APOSTAR em músicas de alta energia para workouts e festas")
    elif "Feliz" in perfil_dominante:
        recomendacoes.append("😊 CRIAR playlists temáticas de músicas positivas para bem-estar")

# Baseado em colaborações
if 'collab' in df.columns and media_collab > media_solo:
    recomendacoes.append("🤝 ESTIMULAR colaborações entre artistas de diferentes países")

# Baseado em estilo latino
if 'is_latin_style' in df.columns and media_latin > media_nao_latin:
    recomendacoes.append("🎸 EXPLORAR fusões de ritmos latinos com gêneros internacionais")

# Baseado em sub-região
if 'sub_region' in df.columns:
    regiao_top = df.groupby('sub_region')['streams_millions'].sum().idxmax()
    recomendacoes.append(f"📍 PRIORIZAR estratégias para {regiao_top}, maior mercado das Américas")

for i, rec in enumerate(recomendacoes, 1):
    print(f"{i}. {rec}")

# ===================================================================
# RESUMO EXECUTIVO FINAL
# ===================================================================

print("\n" + "="*80)
print("📋 RESUMO EXECUTIVO - MERCADO MUSICAL DAS AMÉRICAS")
print("="*80)

print(f"""
┌────────────────────────────────────────────────────────────────────────┐
│                          RESUMO GERAL                                   │
├────────────────────────────────────────────────────────────────────────┤
│ • Total de streams: {total_streams:.1f} milhões                                       │
│ • {df['country'].nunique()} países analisados | {df['artist_names'].nunique():,} artistas | {df['track_name'].nunique():,} músicas    │
│ • Período: {df['week'].min().date()} a {df['week'].max().date()}                                │
├────────────────────────────────────────────────────────────────────────┤
│                         TOP PERFORMERS                                  │
├────────────────────────────────────────────────────────────────────────┤
│ • País líder: {top_pais.index[0]} ({top_pais.values[0]:.1f}M streams)                              │
│ • Artista líder: {top_artista.index[0][:35]}...                                  │
│ • Música líder: {top_musica.index[0][0][:40]}...                               │
├────────────────────────────────────────────────────────────────────────┤
│                      PERFIL MUSICAL MÉDIO                                │
├────────────────────────────────────────────────────────────────────────┤
│ • 💃 Dançabilidade: {media_dance:.3f} │ ⚡ Energia: {media_energy:.3f} │ 😊 Positividade: {media_valence:.3f} │
├────────────────────────────────────────────────────────────────────────┤
│                      PRINCIPAL INSIGHT                                   │
├────────────────────────────────────────────────────────────────────────┤
│ • {recomendacoes[0] if recomendacoes else 'Diversidade musical é a chave do sucesso nas Américas'}       │
└────────────────────────────────────────────────────────────────────────┘
""")

print("\n✅ Análise de insights concluída!")
print("="*80)
```

```python
# ===================================================================
# SALVAR RESULTADOS DA ANÁLISE - AMÉRICAS
# ===================================================================

print("💾 SALVANDO RESULTADOS DA ANÁLISE DAS AMÉRICAS...")
print("=" * 70)

# Criar pasta específica para resultados das Américas
import os
resultados_americas_dir = '/content/drive/MyDrive/spotify_americas_analise/'
try:
    os.makedirs(resultados_americas_dir, exist_ok=True)
    print(f"📁 Pasta criada: {resultados_americas_dir}")
except:
    print(f"⚠️ Usando pasta raiz do Drive")
    resultados_americas_dir = '/content/drive/MyDrive/'

# ===================================================================
# 1. SALVAR TOP MÚSICAS DAS AMÉRICAS
# ===================================================================

print("\n1️⃣ Salvando top músicas das Américas...")
print("-" * 40)

top_musicas_completo = (df.groupby(['track_name', 'artist_names'])
                        .agg({
                            'streams_millions': 'sum',
                            'rank': 'min',
                            'country': 'nunique'
                        })
                        .rename(columns={'country': 'paises_alcancados'})
                        .sort_values('streams_millions', ascending=False)
                        .head(100)  # Top 100 para análise mais completa
                        .reset_index())

# Adicionar sub-região principal da música
if 'sub_region' in df.columns:
    musica_regiao = df.groupby('track_name')['sub_region'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else 'Múltiplas')
    top_musicas_completo['sub_region_principal'] = top_musicas_completo['track_name'].map(musica_regiao)

top_musicas_completo.to_csv(f'{resultados_americas_dir}top_100_musicas_americas.csv', index=False)
print(f"   ✅ top_100_musicas_americas.csv salvo ({len(top_musicas_completo)} músicas)")

# Mostrar exemplo
print(f"   📊 Exemplo: {top_musicas_completo.iloc[0]['track_name'][:40]}... - {top_musicas_completo.iloc[0]['streams_millions']:.1f}M streams")

# ===================================================================
# 2. SALVAR ANÁLISE POR PAÍS DAS AMÉRICAS
# ===================================================================

print("\n2️⃣ Salvando análise por país das Américas...")
print("-" * 40)

analise_paises = (df.groupby('country')
                  .agg({
                      'streams_millions': 'sum',
                      'danceability': 'mean',
                      'energy': 'mean',
                      'valence': 'mean',
                      'track_name': 'count',
                      'artist_names': 'nunique'
                  })
                  .rename(columns={
                      'track_name': 'total_registros',
                      'artist_names': 'total_artistas'
                  })
                  .round(3)
                  .reset_index())

# Ordenar por streams
analise_paises = analise_paises.sort_values('streams_millions', ascending=False)

# Adicionar sub-região
if 'sub_region' in df.columns:
    pais_regiao = df.groupby('country')['sub_region'].first()
    analise_paises['sub_region'] = analise_paises['country'].map(pais_regiao)

analise_paises.to_csv(f'{resultados_americas_dir}analise_por_pais_americas.csv', index=False)
print(f"   ✅ analise_por_pais_americas.csv salvo ({len(analise_paises)} países)")

# Mostrar top 5
print(f"   📊 Top 5 países:")
for _, row in analise_paises.head(5).iterrows():
    print(f"      • {row['country']}: {row['streams_millions']:.1f}M streams")

# ===================================================================
# 3. SALVAR ANÁLISE POR SUB-REGIÃO (se disponível)
# ===================================================================

if 'sub_region' in df.columns:
    print("\n3️⃣ Salvando análise por sub-região...")
    print("-" * 40)

    analise_regioes = (df.groupby('sub_region')
                       .agg({
                           'streams_millions': 'sum',
                           'danceability': 'mean',
                           'energy': 'mean',
                           'valence': 'mean',
                           'track_name': 'count',
                           'country': 'nunique',
                           'artist_names': 'nunique'
                       })
                       .rename(columns={
                           'track_name': 'total_musicas',
                           'country': 'total_paises',
                           'artist_names': 'total_artistas'
                       })
                       .round(3)
                       .reset_index())

    analise_regioes = analise_regioes.sort_values('streams_millions', ascending=False)
    analise_regioes.to_csv(f'{resultados_americas_dir}analise_por_sub_regiao_americas.csv', index=False)
    print(f"   ✅ analise_por_sub_regiao_americas.csv salvo ({len(analise_regioes)} regiões)")

# ===================================================================
# 4. SALVAR TOP ARTISTAS DAS AMÉRICAS
# ===================================================================

print("\n4️⃣ Salvando top artistas das Américas...")
print("-" * 40)

top_artistas_completo = (df.groupby('artist_names')
                         .agg({
                             'streams_millions': 'sum',
                             'track_name': 'nunique',
                             'country': 'nunique'
                         })
                         .rename(columns={
                             'track_name': 'musicas_distintas',
                             'country': 'paises_alcancados'
                         })
                         .sort_values('streams_millions', ascending=False)
                         .head(50)
                         .reset_index())

# Adicionar sub-região principal
if 'sub_region' in df.columns:
    artista_regiao = df.groupby('artist_names')['sub_region'].agg(lambda x: x.mode()[0] if len(x.mode()) > 0 else 'N/A')
    top_artistas_completo['sub_region_principal'] = top_artistas_completo['artist_names'].map(artista_regiao)

top_artistas_completo.to_csv(f'{resultados_americas_dir}top_50_artistas_americas.csv', index=False)
print(f"   ✅ top_50_artistas_americas.csv salvo ({len(top_artistas_completo)} artistas)")

# ===================================================================
# 5. SALVAR PERFIL MUSICAL POR PAÍS
# ===================================================================

if 'music_profile' in df.columns:
    print("\n5️⃣ Salvando perfil musical por país...")
    print("-" * 40)

    # Matriz de perfil musical por país (percentuais)
    perfil_pais_matrix = pd.crosstab(df['country'], df['music_profile'], normalize='index') * 100
    perfil_pais_matrix.round(2).to_csv(f'{resultados_americas_dir}perfil_musical_por_pais_americas.csv')
    print(f"   ✅ perfil_musical_por_pais_americas.csv salvo")

    # Perfil geral consolidado
    perfil_geral = df['music_profile'].value_counts(normalize=True) * 100
    perfil_geral.round(2).to_csv(f'{resultados_americas_dir}perfil_musical_geral_americas.csv')
    print(f"   ✅ perfil_musical_geral_americas.csv salvo")

# ===================================================================
# 6. SALVAR METADADOS COMPLETOS EM JSON
# ===================================================================

print("\n6️⃣ Salvando metadados da análise...")
print("-" * 40)

# Calcular métricas adicionais
if 'success_score' in df.columns:
    media_sucesso = df['success_score'].mean()
    max_sucesso = df['success_score'].max()
else:
    media_sucesso = None
    max_sucesso = None

resumo_analise = {
    'projeto': 'Spotify Weekly Top 200 - Análise das Américas',
    'data_analise': pd.Timestamp.now().isoformat(),
    'total_streams_millions': float(total_streams),
    'total_streams_bilhoes': float(total_streams / 1000),
    'total_musicas_unicas': int(df['track_name'].nunique()),
    'total_artistas_unicos': int(df['artist_names'].nunique()),
    'total_paises': int(df['country'].nunique()),
    'periodo_inicio': df['week'].min().isoformat(),
    'periodo_fim': df['week'].max().isoformat(),
    'media_danceability': float(media_dance),
    'media_energy': float(media_energy),
    'media_valence': float(media_valence),
    'media_streams_por_musica': float(media_streams),
    'top_musica': str(top_musica.index[0][0]) if isinstance(top_musica.index[0], tuple) else str(top_musica.index[0]),
    'top_artista': str(top_artista.index[0]),
    'top_pais': str(top_pais.index[0]),
    'perfil_dominante': str(perfil_dominante),
    'media_success_score': float(media_sucesso) if media_sucesso else None,
    'correlacao_rank_streams': float(corr_rank_streams),
    'correlacao_dance_streams': float(df['danceability'].corr(df['streams_millions'])) if 'danceability' in df.columns else None
}

import json
with open(f'{resultados_americas_dir}resumo_analise_americas.json', 'w') as f:
    json.dump(resumo_analise, f, indent=2, ensure_ascii=False)

print(f"   ✅ resumo_analise_americas.json salvo")

# ===================================================================
# 7. SALVAR RELATÓRIO EM FORMATO LEGÍVEL (TXT)
# ===================================================================

print("\n7️⃣ Salvando relatório executivo em TXT...")
print("-" * 40)

with open(f'{resultados_americas_dir}relatorio_americas.txt', 'w', encoding='utf-8') as f:
    f.write("="*80 + "\n")
    f.write("RELATÓRIO DE ANÁLISE - SPOTIFY AMÉRICAS\n")
    f.write("="*80 + "\n\n")

    f.write("📊 MÉTRICAS GLOBAIS:\n")
    f.write(f"   • Total de streams: {total_streams:.1f} milhões\n")
    f.write(f"   • Média por música: {media_streams:.2f}M\n")
    f.write(f"   • Países: {df['country'].nunique()}\n")
    f.write(f"   • Artistas únicos: {df['artist_names'].nunique():,}\n")
    f.write(f"   • Músicas únicas: {df['track_name'].nunique():,}\n\n")

    f.write("🏆 TOP PERFORMERS:\n")
    f.write(f"   • Música: {top_musica.index[0][0] if isinstance(top_musica.index[0], tuple) else top_musica.index[0]} ({top_musica.values[0]:.1f}M)\n")
    f.write(f"   • Artista: {top_artista.index[0]} ({top_artista.values[0]:.1f}M)\n")
    f.write(f"   • País: {top_pais.index[0]} ({top_pais.values[0]:.1f}M)\n\n")

    f.write("🎸 CARACTERÍSTICAS MUSICAIS:\n")
    f.write(f"   • Dançabilidade média: {media_dance:.3f}\n")
    f.write(f"   • Energia média: {media_energy:.3f}\n")
    f.write(f"   • Positividade média: {media_valence:.3f}\n\n")

    f.write("🎭 PERFIL DOMINANTE:\n")
    f.write(f"   • {perfil_dominante} ({pct_perfil_dominante:.1f}%)\n\n")

    f.write("📈 CORRELAÇÕES:\n")
    f.write(f"   • Rank vs Streams: {corr_rank_streams:.3f}\n")
    f.write(f"   • Dançabilidade vs Streams: {df['danceability'].corr(df['streams_millions']):.3f}\n\n")

    f.write("="*80 + "\n")
    f.write("Fim do relatório\n")

print(f"   ✅ relatorio_americas.txt salvo")

# ===================================================================
# 8. LISTAR TODOS OS ARQUIVOS SALVOS
# ===================================================================

print("\n" + "="*70)
print("📁 ARQUIVOS SALVOS COM SUCESSO!")
print("="*70)

try:
    arquivos = os.listdir(resultados_americas_dir)
    print(f"\n📂 Localização: {resultados_americas_dir}")
    print("-" * 50)

    for arquivo in sorted(arquivos):
        if arquivo.endswith(('.csv', '.json', '.txt')):
            tamanho = os.path.getsize(f"{resultados_americas_dir}{arquivo}")
            if tamanho < 1024:
                tamanho_str = f"{tamanho} B"
            elif tamanho < 1024**2:
                tamanho_str = f"{tamanho/1024:.1f} KB"
            else:
                tamanho_str = f"{tamanho/1024**2:.1f} MB"
            print(f"   📄 {arquivo:45} : {tamanho_str}")
except Exception as e:
    print(f"   ⚠️ Erro ao listar arquivos: {e}")

# ===================================================================
# 9. RESUMO FINAL
# ===================================================================

print("\n" + "="*70)
print("✅ RESULTADOS SALVOS COM SUCESSO!")
print("="*70)

print(f"""
📊 RESUMO DO QUE FOI SALVO:

   ✅ DADOS ANALÍTICOS:
      • top_100_musicas_americas.csv - Top 100 músicas
      • top_50_artistas_americas.csv - Top 50 artistas
      • analise_por_pais_americas.csv - Métricas por país
      • analise_por_sub_regiao_americas.csv - Métricas por sub-região
      • perfil_musical_por_pais_americas.csv - Perfil por país
      • perfil_musical_geral_americas.csv - Perfil geral

   ✅ DOCUMENTAÇÃO:
      • resumo_analise_americas.json - Metadados completos
      • relatorio_americas.txt - Relatório executivo

📂 LOCALIZAÇÃO: {resultados_americas_dir}

💡 DICA: Use pd.read_csv() para carregar os arquivos CSV
   ou pd.read_json() para o arquivo JSON.
""")

print("🎉 Análise das Américas concluída e resultados salvos!")
print("="*70)
```
